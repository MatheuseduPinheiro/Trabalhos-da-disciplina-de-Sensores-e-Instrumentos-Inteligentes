#### Importação das Bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
```

#### Setando o dado de leitura 

```python
dfteste = pd.read_csv('Untitled Experiment Untitled Sensor Preview 1.csv')
```

#### Calcular médias móveis para cada eixo de aceleração

```python
for coluna in ['AccX', 'AccY', 'AccZ']:
    dfteste[f'{coluna}_movel_5'] = dfteste[coluna].rolling(window=5).mean()
    dfteste[f'{coluna}_movel_10'] = dfteste[coluna].rolling(window=10).mean()
    dfteste[f'{coluna}_movel_20'] = dfteste[coluna].rolling(window=20).mean()
```

#### Calculando medianas móveis para cada eixo da aceleração

```python
for coluna in ['AccX', 'AccY', 'AccZ']:
    dfteste[f'{coluna}_mediana_5'] = dfteste[coluna].rolling(window=5).median()
    dfteste[f'{coluna}_mediana_10'] = dfteste[coluna].rolling(window=10).median()
    dfteste[f'{coluna}_mediana_20'] = dfteste[coluna].rolling(window=20).median()
```

#### Setando os Indices

```python
eixos = ['AccX', 'AccY', 'AccZ']
nomes_eixos = ['Aceleração X', 'Aceleração Y', 'Aceleração Z']
```

#### Plotando médias móveis

```python
for i, (eixo, nome) in enumerate(zip(eixos, nomes_eixos)):
    plt.figure(figsize=(12, 8))
    
    plt.subplot(2, 1, 1)
    plt.plot(dfteste[eixo], label=f'{nome} Original', alpha=0.7)
    plt.plot(dfteste[f'{eixo}_movel_5'], label='Média Móvel 5', linewidth=2)
    plt.plot(dfteste[f'{eixo}_movel_10'], label='Média Móvel 10', linewidth=2)
    plt.plot(dfteste[f'{eixo}_movel_20'], label='Média Móvel 20', linewidth=2)
    plt.title(f'Médias Móveis - {nome}')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```

#### Plotando medianas móveis

```python
for i, (eixo, nome) in enumerate(zip(eixos, nomes_eixos)):
    plt.figure(figsize=(12, 8))
    
    plt.subplot(2, 1, 2)
    plt.plot(dfteste[eixo], label=f'{nome} Original', alpha=0.7)
    plt.plot(dfteste[f'{eixo}_mediana_5'], label='Mediana Móvel 5', linewidth=2)
    plt.plot(dfteste[f'{eixo}_mediana_10'], label='Mediana Móvel 10', linewidth=2)
    plt.plot(dfteste[f'{eixo}_mediana_20'], label='Mediana Móvel 20', linewidth=2)
    plt.title(f'Medianas Móveis - {nome}')
    plt.legend()
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()
```
